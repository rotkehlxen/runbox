```python
from getpass import getpass
from pathlib import Path

import garminconnect
from garminconnect import (
    GarminConnectAuthenticationError,
    GarminConnectConnectionError,
    GarminConnectTooManyRequestsError,
)
```

```python
tokenstore = str(Path("~/.garminconnect").expanduser())

try:
    garmin = garminconnect.Garmin()
    garmin.login(tokenstore)
    print(f"✅ Logged in with stored tokens from: {tokenstore}")

except (
    FileNotFoundError,
    GarminConnectAuthenticationError,
    GarminConnectConnectionError,
) as err:
    print(f"Stored token login failed: {err}")
    print("Requesting credentials to create fresh tokens...")

    email = input("Enter email address: ").strip()
    password = getpass("Enter password: ")

    garmin = garminconnect.Garmin(email=email, password=password)
    garmin.login()
    garmin.client.dump(tokenstore)
    print(f"✅ New tokens created and saved to: {tokenstore}")

except GarminConnectTooManyRequestsError as err:
    print(f"❌ Garmin rate limit hit: {err}")
    raise
```
